```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error


df_ml=pd.get_dummies(df[['GarmentType', 'Material', 'Cost']], drop_first='True')


X=df_ml.drop('Cost', axis=1)
Y=df_ml['Cost']

X_train, X_test, Y_train, Y_test=train_test_split(X,Y, test_size=0.2, random_state=42)

model=RandomForestRegressor(random_state=42)
model.fit(X_train, Y_train)

predictions=model.predict(X_test)

error=mean_absolute_error(Y_test, predictions)

print(f'Modelning o\'rtacha xatoligi (MAE): ${error:.2f}')
```

```python
features=['GarmentType', 'Material', 'Chest', 'Waist', 'Hip', 'Shoulder', 'Arm Length']

df_ml=pd.get_dummies(df[features + ['Cost']], drop_first=True)

X=df_ml.drop('Cost', axis=1)
Y=df_ml['Cost']

X_train, X_test, Y_train, Y_test=train_test_split(X,Y,test_size=0.2,random_state=42)

model=RandomForestRegressor(random_state=42)
model.fit(X_train, Y_train)
predictions=model.predict(X_test)
error=mean_absolute_error(Y_test, predictions)
print(f'Keyingi Xatolik (MAE): ${error:.2f}')
```

```python
import matplotlib.pyplot as plt

monthly_cost=df.groupby(df['Date'].dt.month)['Cost'].sum()
plt.figure(figsize=(10, 5))
plt.plot(monthly_cost.index, monthly_cost.values, marker='o', color='#2b5c8f', linewidth=2.5, markersize=8)

plt.title('Monthy Order Cost Trend', fontsize=14, fontweight='bold')
plt.xlabel('Month (1-January, 12- December)', fontsize=11)
plt.ylabel('Total Cost ($)',fontsize=11)
plt.grid(True, linestyle='--', alpha=0.6)

plt.show()
```
